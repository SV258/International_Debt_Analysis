```python
cursor = connection.cursor()

cursor.execute("SHOW CREATE TABLE countries")
print(cursor.fetchone()[1])

cursor.execute("SHOW CREATE TABLE indicators")
print(cursor.fetchone()[1])
```

```python
cursor.execute("""
CREATE TABLE IF NOT EXISTS debt_observations (
    country_code VARCHAR(10) NOT NULL,
    indicator_code VARCHAR(50) NOT NULL,
    year INT NOT NULL,
    debt_value DOUBLE NOT NULL,

    PRIMARY KEY (country_code, indicator_code, year),

    FOREIGN KEY (country_code)
        REFERENCES countries(country_code),

    FOREIGN KEY (indicator_code)
        REFERENCES indicators(indicator_code)
) ENGINE=InnoDB
  DEFAULT CHARSET=utf8mb4
  COLLATE=utf8mb4_0900_ai_ci
""")

print("debt_observations table ready")
```

```python
insert_query = """
INSERT INTO debt_observations
    (country_code, indicator_code, year, debt_value)
VALUES (%s, %s, %s, %s)
ON DUPLICATE KEY UPDATE
    debt_value = debt_observations.debt_value
"""

first_batch = list(
    debt_all_sql[
        ["country_code", "indicator_code", "year", "debt_value"]
    ].head(1000).itertuples(index=False, name=None)
)

try:
    cursor.executemany(insert_query, first_batch)
    connection.commit()
    print("First batch saved successfully")
except Exception:
    connection.rollback()
    raise
```

```python
columns = [
    "country_code", "indicator_code", "year", "debt_value"
]
batch_size = 5000
total = len(debt_all_sql)

for start in range(0, total, batch_size):
    batch = list(
        debt_all_sql[columns]
        .iloc[start:start + batch_size]
        .itertuples(index=False, name=None)
    )

    try:
        cursor.executemany(insert_query, batch)
        connection.commit()
    except Exception:
        connection.rollback()
        raise

    processed = min(start + batch_size, total)

    if processed % 50000 == 0 or processed == total:
        print(f"Processed {processed:,} of {total:,} rows")

print("Data loading completed")
```

## EDA Findings

- In 2024, total external debt across the 120 countries in
  this dataset was approximately $8.94 trillion.

- China had the highest external debt at $2.42 trillion,
  accounting for 27.07% of the dataset total.

- India ranked second with $716 billion, followed by
  Brazil with $605 billion.

- China, India, Brazil, Mexico and Türkiye each accounted
  for more than 5% of the dataset's 2024 total.

- Reported annual debt totals increased overall from
  approximately $1.83 trillion in 2000 to $8.94 trillion
  in 2024. Country coverage varies by year, so this is
  not a comparison of an identical country group.

## Interpretation Notes

- Main debt totals use indicator DT.DOD.DECT.CD.
- Missing observations are excluded, not treated as zero.
- Different debt stock indicators can overlap and should
  not be added together.
- Values use current USD and are not adjusted for inflation.
- Debt amounts alone do not establish a country's debt risk.
